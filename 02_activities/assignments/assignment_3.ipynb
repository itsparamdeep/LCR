```python
# Import standard libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

#### **Question 2:** 
#### Data-visualization

Let's create plots to visualize the relationships between the features (e.g., alcohol, malic acid, ash...).


```python
def plot_feature_pairs(data, feature_names, color_labels=None, title_prefix=''):
    """
    Helper function to create scatter plots for all possible pairs of features.
    
    Parameters:
    - data: DataFrame containing the features to be plotted.
    - feature_names: List of feature names to be used in plotting.
    - color_labels: Optional. Cluster or class labels to color the scatter plots.
    - title_prefix: Optional. Prefix for plot titles to distinguish between different sets of plots.
    """
    # Create a figure for the scatter plots
    plt.figure(figsize=(60, 60))
    
    # Counter for subplot index
    plot_number = 1
    
    # Loop through each pair of features
    for i in range(len(feature_names)):
        for j in range(i + 1, len(feature_names)):
            plt.subplot(len(feature_names)-1, len(feature_names)-1, plot_number)
            
            # Scatter plot colored by labels if provided
            if color_labels is not None:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], 
                            c=color_labels, cmap='viridis', alpha=0.7)
            else:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], alpha=0.7)
            
            plt.xlabel(feature_names[i])
            plt.ylabel(feature_names[j])
            plt.title(f'{title_prefix}{feature_names[i]} vs {feature_names[j]}')
            
            # Increment the plot number
            plot_number += 1

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Show the plot
    plt.show()

# Get feature names
feature_names = wine_df.columns

# Use the helper function to plot scatter plots without coloring by cluster labels
plot_feature_pairs(wine_df, feature_names, title_prefix='Original Data: ')
```

**Question:**
- Do you notice any patterns or relationships between the different features? How might these patterns help in distinguishing between different species?

Yes, a few features move together, like flavanoids, phenols, and color intensity tend to increase together, while hue tends to go down as color intensity goes up. When we look at these pairs, each wine type clusters in a different area rather than mixing randomly, so features like flavanoids and color intensity are pretty good at telling the species apart. Features like ash and magnesium, though, overlap a lot across all three types and don't help much with distinguishing them.

```python
# Initialize the StandardScaler
scaler = StandardScaler()

# Scale all the features in the dataset
scaled_features = scaler.fit_transform(wine_df)

# Create a new DataFrame with scaled features
scaled_wine_df = pd.DataFrame(scaled_features, columns=wine_data.feature_names)

# Display the first few rows of the scaled DataFrame
print(scaled_wine_df.head())
```

Why is it important to standardize the features of a dataset before applying clustering algorithms like K-Means? Discuss the implications of using unstandardized data in your analysis.

>Standardization matters because K-Means groups points based on distance, and distance is scale-dependent, so a feature with a much bigger range, like proline , would completely dominate over a feature like hue , even if hue is just as important for finding real groups. Without standardizing, clustering would basically end up being driven by whichever feature has the largest numbers, rather than reflecting all the features together, and could produce clusters that don't match any real underlying grouping like the wine cultivars. Standardizing puts every feature on the same scale, so each one contributes fairly to the distance calculation, which is why it's a necessary step before running K-Means rather than an optional one.

#### **Question 4:** 
#### K-means clustering 

Apply the K-Means clustering algorithm to the Wine dataset. Choose the value 3 for the number of clusters (`k=3`) and fit the model. Assign cluster labels to the original data and add them as a new column in the DataFrame.

```python
# Initialize the KMeans model with 3 clusters
kmeans = KMeans(n_clusters=3, random_state=42)

# Fit the model to the wine data
kmeans.fit(wine_df)

clustered_wine_data = wine_df.copy()
clustered_wine_data['Cluster'] = kmeans.labels_


# Use the helper function to plot scatter plots, colored by cluster labels
plot_feature_pairs(clustered_wine_data, feature_names, color_labels=clustered_wine_data['Cluster'], title_prefix='Clustered Data: ')
```

We chose `k=3` for the number of clusters arbitrarily. However, in a real-world scenario, it is important to determine the optimal number of clusters using appropriate methods.

#### **Question 5:** 
#### Bootstrapping 

Implement bootstrapping on the mean of `color intensity`. Generate 10000 bootstrap samples, calculate the mean for each sample, and compute a 90% confidence interval.

```python
# Your answer here...

mean_color_intensity = clustered_wine_data['color_intensity'].mean()

np.random.seed(123)

n_bootstrap_samples = 1000
n = len(clustered_wine_data)
boot_means = np.empty(n_bootstrap_samples)

for i in range(n_bootstrap_samples):
    # Generate a bootstrap sample
    bootstrap_sample = clustered_wine_data['color_intensity'].sample(n=n, replace=True)
    # Calculate the mean of the bootstrap sample
    boot_means[i] = bootstrap_sample.mean()

lower_bound = np.percentile(boot_means, 5)
upper_bound = np.percentile(boot_means, 95) 

# Display the result
print(f"Mean of Color Intensity: {mean_color_intensity}")
print(f"90% Confidence Interval of Mean Color Intensity: ({lower_bound}, {upper_bound})")
```

**Question:**
- Why do we use bootstrapping in this context? What does it help us understand about the mean?

>  We use bootstrapping because we only have one sample, and resampling from it with replacement lets us estimate how much the mean would vary if we had drawn different samples, without needing more real data or assuming a specific distribution. It helps us understand not just what the mean is, but how confident we can be in the estimate, by showing the range of plausible values the true mean could fall in 

**Question:**
- What is the purpose of calculating the confidence interval from the bootstrap samples? How does it help us interpret the variability of the estimate?

> The confidence interval turns the single mean estimate into a range of plausible values, so instead of just knowing the mean is around 5.058, we know it likely falls between about 4.77 and 5.35. This helps us interpret variability because a narrow interval means the estimate is fairly reliable, while a wider interval means there's more uncertainty, giving us a sense of how much to trust the estimate rather than treating it as an exact number.

**Question:**

- Reflect on the variability observed in the bootstrapped means and discuss whether the mean of the color intensity appears to be a stable and reliable estimate based on the confidence interval and the spread of the bootstrapped means.

> The bootstrapped means cluster tightly around 5.058, with a narrow 90% confidence interval of about (4.77, 5.35), even though the raw color_intensity values themselves range much more widely, from about 1.3 to 13. This suggests the mean is a fairly stable and reliable estimate, since it isn't likely to shift much if we drew a different sample of wines, even though individual wines vary a lot.
